```python
import pandas as pd
df = pd.read_excel("Engine Draft.xlsx")
```

```python
print(df.columns)
```

```python
def merge_engines(row):
    e1 = str(row["Engine1"]).split(",") if pd.notna(row["Engine1"]) else []
    e2 = str(row["Engine2"]).split(",") if pd.notna(row["Engine2"]) else []

    # strip spaces and preserve order
    merged = []
    for e in e1 + e2:
        e = e.strip()
        if e and e not in merged:
            merged.append(e)

    return ",".join(merged)

df["Engine"] = df.apply(merge_engines, axis=1)

print(df[["Engine1", "Engine2", "Engine"]].head())
```
