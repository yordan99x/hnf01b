```python
import pandas as pd
import numpy as np
# Read Excel
df = pd.read_excel("engine type.xlsx",sheet_name="NESEA - New - 2023")
```

```python
def normalize_disp(d):
    if pd.isna(d):
        return None
    return str(d).strip()

def classify_engine(row):
    model = str(row["Engine Model"]).strip().upper()
    disp = normalize_disp(row["Displacement"])

    # --- 855 ---
    if model == "855" and disp == "14":
        return "NT855"
    elif model == "855" and disp == "12":
        return "NT855"
    elif model == "855" and disp == "19":
        return "KTA19"
    
    # --- A SERIES ---
    elif model == "A" and disp == "1.7":
        return "A1.7"

    # --- B SERIES ---
    elif model == "B" and disp == "2.8":
        return "4BT2.8"
    elif model == "B" and disp == "3.3":
        return "4BT3.3"
    elif model == "B" and disp == "3.8":
        return "4BT3.8"
    elif model == "B" and disp == "3.9":
        return "4BT3.9"
    elif model == "B" and disp == "4.5":
        return "6BT4.5"
    elif model == "B" and disp == "5.9":
        return "6BT5.9"
    elif model == "B" and disp == "6.2":
        return "6BT6.2"
    elif model == "B" and disp == "6.7":
        return "6BT6.7"
    elif model == "B" and disp == "8.3":
        return "6BT8.3"
    elif model == "B" and disp == "8.9":
        return "6BT8.9"
    elif model == "B" and disp == "9.5":
        return "6BT9.5"
    elif model == "B" and disp == "903":
        return "V903"


    # --- C ---
    elif model == "C" and disp == "5.9":
        return "6CT5.9"
    elif model == "C" and disp == "8.3":
        return "6CT8.3"
    elif model == "C" and disp == "8.9":
        return "6CT8.3"
    elif model  == "C" and disp == "9.5":
        return "6CT9.5"
    
    # --- C/L ---
    elif model == "C/L" and disp == "8.9":
        return "L8.9"
    elif model == "C/L" and disp == "9.3":
        return "L9.3"
    elif model == "C/L" and disp == "9.5":
        return "L9.5"
    
    # --- H ---
    elif model == "H" and disp == "78":
        return "QSK78"
    # --- HSK ---
    elif model == "HSK" and disp == "78":
        return "QSK78 Gas"
    
    # --- ISB/QSB ---
    elif model == "ISB/QSB" and disp == "2.8":
        return "ISB2.8, QSB2.8"
    elif model == "ISB/QSB" and disp == "3.3":
        return "ISB3.3, QSB3.3"
    elif model == "ISB/QSB" and disp == "3.8":
        return "ISB3.8, QSB3.8"
    elif model == "ISB/QSB" and disp == "3.9":
        return "ISB3.9, QSB3.9"
    elif model == "ISB/QSB" and disp == "4.5":
        return "ISB4.5, QSB4.5"
    elif model == "ISB/QSB" and disp == "5.0":
        return "ISB5.0, QSB5.0"
    elif model == "ISB/QSB" and disp == "5.9":
        return "ISB5.9, QSB5.9"
    elif model == "ISB/QSB" and disp == "6.2":
        return "ISB6.2, QSB6.2"
    elif model == "ISB/QSB" and disp == "6.7":
        return "ISB6.7, QSB6.7"
    elif model == "ISB/QSB" and disp == "7":
        return "ISB7, QSB7"
    elif model == "ISB/QSB" and disp == "8.3":
        return "ISB8.3, QSB8.3"
    elif model == "ISB/QSB" and disp == "8.9":
        return "ISB8.9, QSB8.9"
    elif model == "ISB/QSB" and disp == "11":
        return "ISB11, QSB11"
    elif model == "ISB/QSB" and disp == "12":
        return "ISB12, QSB12"
    elif model == "ISB/QSB" and disp == "14":
        return "ISB14, QSB14"
    elif model == "ISB/QSB" and disp == "15":
        return "ISB15, QSB15"
    elif model == "ISB/QSB" and disp == "30":
        return "ISB30, QSB30"
    elif model == "ISB/QSB" and disp == "60":
        return "ISB60, QSB60"
    
    # --- ISC/ISL ---
    elif model == "ISC/ISL" and disp == "3.9":
        return "ISC3.9, ISL3.9"
    elif model == "ISC/ISL" and disp == "8.3":
        return "ISC8.3, ISL8.3"
    elif model == "ISC/ISL" and disp == "8.4":
        return "ISC8.4, ISL8.4"
    elif model == "ISC/ISL" and disp == "8.9":
        return "ISC8.9, ISL8.9"
    elif model == "ISC/ISL" and disp == "9.3":
        return "ISC9.3, ISL9.3"
    elif model == "ISC/ISL" and disp == "9.5":
        return "ISC9.5, ISL9.5"
    
    # --- ISF/QSF
    elif model == "ISF/QSF" and disp == "2.5":
        return "ISF2.5, QSF2.5"
    elif model == "ISF/QSF" and disp == "2.8":
        return "ISF2.8, QSF2.8"
    elif model == "ISF/QSF" and disp == "3.8":
        return "ISF3.8, QSF3.8"
    elif model == "ISF/QSF" and disp == "3.9":
        return "ISF3.9, QSF3.9"
    elif model == "ISF/QSF" and disp == "4":
        return "ISF4, QSF4"
    elif model == "ISF/QSF" and disp == "4.5":
        return "ISF4.5, QSF4.5"
    elif model == "ISF/QSF" and disp == "5.0":
        return "ISF5.0, QSF5.0"
    elif model == "ISF/QSF" and disp == "5.9":
        return "ISF5.9, QSF5.9"
    elif model == "ISF/QSF" and disp == "6.7":
        return "ISF6.7, QSF6.7"
    elif model == "ISF/QSF" and disp == "8.9":
        return "ISF8.9, QSF8.9"
    elif model == "ISF/QSF" and disp == "12":
        return "ISF12, QSF12"
    
    # --- ISG / QSG ---
    elif model == "ISG/QSG" and disp == "11":
        return "ISG11, QSG11"
    elif model == "ISG/QSG" and disp == "12":
        return "ISG12, QSG12"
    elif model == "ISG/QSG" and disp == "12/15":
        return "ISG12, ISG15, QSG12, QSG15"
    elif model == "ISG/QSG" and disp == "13":
        return "ISG13, QSG13"
    elif model == "ISG/QSG" and disp == "23":
        return "ISG23, QSG23"
    elif model == "ISG/QSG" and disp == "8.3":
        return "ISG8.3, QSG8.3"
    
    # -- ISM/QSM ---
    elif model == "ISM/QSM" and disp == "11":
        return "ISM11, QSM11"
    elif model == "ISM/QSM" and disp == "12":
        return "ISM12, QSM12"
    elif model == "ISM/QSM" and disp == "14":
        return "ISM14, QSM14"
    
    # --- ISN/QSN ---
    elif model == "ISN/QSN" and disp == "14":
        return "ISN14, QSN14"
    
    # --- ISV ---
    elif model == "ISV" and disp == "5.0":
        return "ISV5.0"

    # --- ISX / QSX ---
    elif model == "ISX/QSX" and disp == "11":
        return "ISX11, QSX11"
    elif model == "ISX/QSX" and disp == "12":
        return "ISX12, QSX12"
    elif model == "ISX/QSX" and disp == "12/15":
        return "ISX12, ISX15, QSX12, QSX15"
    elif model == "ISX/QSX" and disp == "13":
        return "ISX13, QSX13"
    elif model == "ISX/QSX" and disp == "15":
        return "ISX15, QSX15"
    elif model == "ISX/QSX" and disp == "17":
        return "ISX17, QSX17"
    elif model == "ISX/QSX" and disp == "30":
        return "ISX30, QSX30"
    elif model == "ISX/QSX" and disp == "8.9":
        return "ISX8.9, QSX8.9"
    
    # --- ISZ ---
    elif model == "ISZ" and disp == "12":
        return "ISZ12"
    elif model == "ISZ" and disp == "13":
        return "ISZ13"
    elif model == "ISZ" and disp == "14":
        return "ISZ14"
    elif model == "ISZ" and disp == "15":
        return "ISZ15"

    # --- K & K / QSK ---
    elif model == "K" and disp == "14":
        return "K14"
    elif model == "K" and disp == "19":
        return "K19"
    elif model == "K" and disp == "19/38/50":
        return "K19, K38, K50"
    elif model == "K" and disp == "38":
        return "K38"
    elif model == "K" and disp == "38/50":
        return "K38, K50"
    elif model == "K" and disp == "50":
        return "K50"
    elif model == "K" and disp == "60":
        return "K60"
    elif model == "K/QSK" and disp == "14":
        return "K14, QSK14"
    elif model == "K/QSK" and disp == "19":
        return "K19, QSK19"
    elif model == "K/QSK" and disp == "19/38/50":
        return "K19, QSK19, K38, QSK38, K50, QSK50"
    elif model == "K/QSK" and disp == "38":
        return "K38, QSK38"
    elif model == "K/QSK" and disp == "38/50":
        return "K38, QSK38, K50, QSK50"
    elif model == "K/QSK" and disp == "50":
        return "K50, QSK50"
    elif model == "K/QSK" and disp == "95":
        return "K95, QSK95"
    
    # --- L10 ---
    elif model == "L10" and disp == "10":
        return "L10"
    elif model == "L10" and disp == "11":
        return "L10"
    elif model == "L10" and disp == "10/11":
        return "L10"
    
    # --- M/L ---
    elif model == "M/L" and disp == "10/11":
        return "L10, M11"
    elif model == "M/L" and disp == "11":
        return "L10, M11"
    
    # --- M11 ---
    elif model == "M11" and disp == "11":
        return "M11"
    elif model == "M11" and disp == "10/11":
        return "M11"
    
    # --- N ---
    elif model == "N14" and disp == "14":
        return "N14"
    elif model == "N14" and disp == "14.8":
        return "N14"
    
    # --- NT/NH ---
    elif model == "NT/NH" and disp == "14":
        return "N14"
    
    # --- N743 ---
    elif model == "N743" and disp == "0":
        return "N743 India"
    
    # --- QSK ---
    elif model == "QSK" and disp == "19":
        return "QSK19"
    elif model == "QSK" and disp == "19/38/50":
        return "QSK19, QSK38, QSK50"
    elif model == "QSK" and disp == "23":
        return "QSK23"
    elif model == "QSK" and disp == "38":
        return "QSK38"
    elif model == "QSK" and disp == "38/50":
        return "QSK38, QSK50"
    elif model == "QSK" and disp == "45":
        return "QSK45"
    elif model == "QSK" and disp == "45/60":
        return "QSK45, QSK60"
    elif model == "QSK" and disp == "50":
        return "QSK50"
    elif model == "QSK" and disp == "60":
        return "QSK60"
    elif model == "QSK" and disp == "78":
        return "QSK78"
    elif model == "QSK" and disp == "95":
        return "QSK95"
    
    # --- QST ---
    elif model == "QST" and disp == "30":
        return "QST30"
    
    # --- QSV ---
    elif model == "QSV" and disp == "73":
        return "QSV73"
    elif model == "QSV" and disp == "81":
        return "QSV81"
    elif model == "QSV" and disp == "91":
        return "QSV91"
    
    # --- QSW ---
    elif model == "QSW" and disp == "73":
        return "QSW73"
    elif model == "QSW" and disp == "82":
        return "QSW82"
    
    
    
    # --- ECJV ---
    elif model == "Endurant - ECJV" and disp == "0":
        return "Eaton"
    elif model == "Procision - ECJV" and disp == "0":
        return "Eaton"
    
    # --- V ---
    elif model == "V" and disp == "2.8":
        return "QSK19"
    elif model == "V" and disp == "28":
        return "VTA28"
    elif model == "V" and disp == "903":
        return "V903"

    else:
        return np.nan
```
