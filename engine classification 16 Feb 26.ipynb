```python
import pandas as pd
```

```python
df1 = pd.read_excel("Engine model draft 16 Feb 26.xlsx", sheet_name = "Sheet1")
df2 = pd.read_excel("engine modelxlsx.xlsx", sheet_name= "Sheet2")
df3 = pd.read_excel("PMF Where Used.xlsx", sheet_name= "Sheet2")
```

```python
df_all = pd.concat([df1,df2,df3], ignore_index=True)
```

```python
def merge_unique(series):
    engines = []
    for val in series.dropna():
        engines.extend([x.strip() for x in val.split(",")])
        
    engines_unique = list(dict.fromkeys(engines))
    return ", ".join(engines_unique)

df_all_merged = (
    df_all.groupby("PN AL78", as_index = False)["Engine Classification"]
    .apply(merge_unique)
)

print(df_all_merged)
```

```python
df_all_merged.to_excel("Engine Classification 2 Maret 26.xlsx", index=False)
```
